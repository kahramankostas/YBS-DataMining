# 03 - Matplotlib ile Veri Görselleştirme
**Veri Madenciliği Dersi – Gümüşhane Üniversitesi, Yönetim Bilişim Sistemleri**

Görselleştirme, veri madenciliğinin hem **başında** (veriyi tanımak, aykırı değerleri ve ilişkileri görmek) hem de **sonunda** (model sonuçlarını raporlamak) kullanılır.

### Öğrenme Hedefleri
1. `pyplot` arayüzü ile **nesne yönelimli (Figure / Axes)** arayüzün farkını anlamak
2. Temel grafik türlerini **doğru amaçla** seçmek: çizgi, sütun, histogram, saçılım, kutu, pasta
3. Başlık, eksen etiketi, renk, açıklama (legend), not (annotation) ile grafiği okunur kılmak
4. `subplots` ile çoklu grafik panelleri hazırlamak
5. pandas'ın yerleşik çizim desteğini kullanmak, grafikleri dosyaya kaydetmek

> 💡 **Altın kural:** Her grafiğin bir **başlığı**, **eksen etiketleri** ve (birden fazla seri varsa) **açıklaması** olmalı. Okuyucu, grafiği metni okumadan anlayabilmeli.

```python
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt

%matplotlib inline
plt.rcParams["figure.figsize"] = (8, 4.5)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.3
print("Matplotlib sürümü:", matplotlib.__version__)
```

## 1. İki Arayüz: `plt` ve `fig, ax` ⭐
- **pyplot (durum tabanlı):** `plt.plot(...)`, `plt.title(...)` – hızlı denemeler için pratik.
- **Nesne yönelimli:** `fig, ax = plt.subplots()` → `ax.plot(...)`, `ax.set_title(...)` – birden fazla grafik ve ince ayar için **önerilen** yöntem.

Bir grafiğin anatomisi: **Figure** (tüm tuval) → bir veya daha fazla **Axes** (tek bir grafik alanı) → eksenler, çizgiler, etiketler...

```python
aylar = ["Oca", "Şub", "Mar", "Nis", "May", "Haz", "Tem", "Ağu", "Eyl", "Eki", "Kas", "Ara"]
ziyaretci = [820, 760, 950, 1100, 1350, 1600, 2100, 2300, 1500, 1050, 870, 990]

# 1) pyplot arayüzü
plt.plot(aylar, ziyaretci)
plt.title("Gümüşhane Karaca Mağarası Aylık Ziyaretçi (kurgusal)")
plt.show()
```

```python
# 2) Nesne yönelimli arayüz (aynı grafik)
fig, ax = plt.subplots()
ax.plot(aylar, ziyaretci)
ax.set_title("Gümüşhane Karaca Mağarası Aylık Ziyaretçi (kurgusal)")
ax.set_xlabel("Ay")
ax.set_ylabel("Ziyaretçi sayısı")
plt.show()
```

## 2. Çizgi Grafiği – Zaman İçindeki Değişim
**Ne zaman?** Sıralı (genellikle zamana bağlı) verilerde eğilim (trend) göstermek için.

```python
gecen_yil = [700, 690, 880, 1000, 1200, 1450, 1900, 2050, 1400, 980, 800, 900]

fig, ax = plt.subplots()
ax.plot(aylar, ziyaretci, marker="o", color="tab:blue", linewidth=2, label="2025")
ax.plot(aylar, gecen_yil, marker="s", color="tab:gray", linestyle="--", label="2024")

# En yüksek noktayı işaretle (annotation)
i = int(np.argmax(ziyaretci))
ax.annotate(f"Zirve: {ziyaretci[i]}", xy=(i, ziyaretci[i]), xytext=(i - 4, ziyaretci[i] - 150),
            arrowprops=dict(arrowstyle="->", color="black"))

ax.set_title("Aylık Ziyaretçi Karşılaştırması")
ax.set_xlabel("Ay")
ax.set_ylabel("Ziyaretçi sayısı")
ax.legend()
plt.show()
```

> **Biçim kısayolları:** `color` (renk), `linestyle` (`"-"`, `"--"`, `":"`, `"-."`), `marker` (`"o"`, `"s"`, `"^"`, `"x"`), `linewidth`, `alpha` (saydamlık 0–1).

## 3. Sütun (Bar) Grafiği – Kategorileri Karşılaştırma
**Ne zaman?** Kategorik değişkenlerin değerlerini karşılaştırmak için. Kategoriler çok veya isimleri uzunsa **yatay** (`barh`) tercih edin.

```python
kategoriler = ["Elektronik", "Giyim", "Kitap", "Kırtasiye", "Ev"]
ciro = [482, 215, 38, 22, 160]   # bin TL

fig, ax = plt.subplots()
cubuklar = ax.bar(kategoriler, ciro, color=["tab:blue", "tab:orange", "tab:green", "tab:red", "tab:purple"])
ax.bar_label(cubuklar, fmt="%d bin ₺", padding=3)   # değerleri çubukların üstüne yaz
ax.set_title("Kategorilere Göre Yıllık Ciro")
ax.set_ylabel("Ciro (bin TL)")
ax.set_ylim(0, max(ciro) * 1.15)
plt.show()
```

```python
# Gruplanmış sütun grafiği: iki şubeyi çeyrek bazında karşılaştırma
ceyrekler = ["Ç1", "Ç2", "Ç3", "Ç4"]
gumushane = [2.1, 2.5, 3.0, 3.8]
trabzon = [4.2, 4.0, 4.9, 5.5]

x = np.arange(len(ceyrekler))
genislik = 0.38

fig, ax = plt.subplots()
ax.bar(x - genislik / 2, gumushane, genislik, label="Gümüşhane")
ax.bar(x + genislik / 2, trabzon, genislik, label="Trabzon")
ax.set_xticks(x, ceyrekler)
ax.set_title("Şubelere Göre Çeyreklik Ciro")
ax.set_ylabel("Milyon TL")
ax.legend()
plt.show()
```

```python
# Yatay, sıralı sütun grafiği (sıralama okunabilirliği çok artırır)
iller = ["Bayburt", "Gümüşhane", "Artvin", "Rize", "Giresun", "Ordu", "Trabzon"]
ogrenci = [9, 14, 11, 18, 21, 25, 47]   # kurgusal: bölüme gelen öğrenci sayısı

sira = np.argsort(ogrenci)
fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(np.array(iller)[sira], np.array(ogrenci)[sira], color="teal")
ax.set_title("Bölüme Gelen Öğrencilerin İllere Göre Dağılımı")
ax.set_xlabel("Öğrenci sayısı")
ax.grid(axis="y", visible=False)
plt.show()
```

## 4. Histogram – Bir Değişkenin Dağılımı ⭐
**Ne zaman?** Sayısal bir değişkenin nasıl dağıldığını (simetrik mi, çarpık mı, kaç tepeli?) görmek için. Veri madenciliğinde **ilk bakılan grafiklerden biridir.**

```python
rng = np.random.default_rng(42)
notlar = np.clip(rng.normal(62, 15, 400), 0, 100)
gelir = rng.lognormal(mean=10.3, sigma=0.5, size=400)   # sağa çarpık dağılım

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(notlar, bins=20, color="steelblue", edgecolor="white")
axes[0].axvline(notlar.mean(), color="red", linestyle="--", label=f"Ortalama = {notlar.mean():.1f}")
axes[0].set_title("Sınav Notları (yaklaşık simetrik)")
axes[0].set_xlabel("Not")
axes[0].set_ylabel("Öğrenci sayısı")
axes[0].legend()

axes[1].hist(gelir, bins=30, color="darkorange", edgecolor="white")
axes[1].axvline(np.mean(gelir), color="red", linestyle="--", label=f"Ortalama = {np.mean(gelir):,.0f}")
axes[1].axvline(np.median(gelir), color="green", linestyle="-", label=f"Medyan = {np.median(gelir):,.0f}")
axes[1].set_title("Aylık Gelir (sağa çarpık)")
axes[1].set_xlabel("TL")
axes[1].legend()

plt.tight_layout()
plt.show()
```

> 🔎 **Yorum:** Sağa çarpık dağılımlarda ortalama, medyandan büyüktür; çünkü birkaç yüksek değer ortalamayı yukarı çeker. Bu tür değişkenlerde medyan daha güvenilir bir merkez ölçüsüdür. `bins` sayısını değiştirerek grafiğin nasıl değiştiğini deneyin.

## 5. Kutu Grafiği (Boxplot) – Dağılım ve Aykırı Değerler ⭐
Kutu: Q1–Q3 aralığı (IQR), içindeki çizgi: medyan, bıyıklar: 1.5×IQR sınırı, noktalar: **aykırı değerler**. Grupları karşılaştırmak için idealdir.

```python
ybs = rng.normal(68, 12, 80)
isletme = rng.normal(61, 15, 90)
iktisat = np.append(rng.normal(58, 10, 70), [5, 8, 99])   # birkaç aykırı değer ekledik

fig, ax = plt.subplots()
ax.boxplot([ybs, isletme, iktisat], tick_labels=["YBS", "İşletme", "İktisat"], patch_artist=True,
           boxprops=dict(facecolor="lightblue"))
ax.set_title("Bölümlere Göre Veri Madenciliği Dersi Notları")
ax.set_ylabel("Not")
plt.show()
```

## 6. Saçılım (Scatter) Grafiği – İki Değişken Arasındaki İlişki ⭐
**Ne zaman?** İki sayısal değişken arasında ilişki (korelasyon), kümeler veya aykırı değerler aramak için.

```python
calisma_saati = rng.uniform(0, 20, 120)
sinav_notu = np.clip(35 + 3 * calisma_saati + rng.normal(0, 8, 120), 0, 100)
devam = rng.uniform(40, 100, 120)   # derse devam yüzdesi

fig, ax = plt.subplots(figsize=(8, 5))
nokta = ax.scatter(calisma_saati, sinav_notu, c=devam, cmap="viridis", s=40, alpha=0.8, edgecolors="k", linewidths=0.3)
fig.colorbar(nokta, ax=ax, label="Devam oranı (%)")

# Eğilim çizgisi (doğrusal uydurma)
a, b = np.polyfit(calisma_saati, sinav_notu, deg=1)
xs = np.linspace(0, 20, 100)
ax.plot(xs, a * xs + b, color="red", label=f"Eğilim: y = {a:.2f}x + {b:.1f}")

r = np.corrcoef(calisma_saati, sinav_notu)[0, 1]
ax.set_title(f"Haftalık Çalışma Süresi ve Sınav Notu (r = {r:.2f})")
ax.set_xlabel("Haftalık çalışma (saat)")
ax.set_ylabel("Sınav notu")
ax.legend()
plt.show()
```

```python
# Gruplara göre renklendirme: sınıflandırma/kümeleme sonuçlarını göstermenin temel yolu
segmentler = {"Ekonomik": (25, 30000, "tab:green"), "Standart": (40, 55000, "tab:blue"), "Premium": (52, 95000, "tab:red")}

fig, ax = plt.subplots(figsize=(8, 5))
for ad, (yas_ort, gelir_ort, renk) in segmentler.items():
    yas = rng.normal(yas_ort, 5, 60)
    gel = rng.normal(gelir_ort, 8000, 60)
    ax.scatter(yas, gel, label=ad, color=renk, alpha=0.7)

ax.set_title("Müşteri Segmentleri: Yaş ve Gelir")
ax.set_xlabel("Yaş")
ax.set_ylabel("Aylık gelir (TL)")
ax.yaxis.set_major_formatter(matplotlib.ticker.StrMethodFormatter("{x:,.0f}"))
ax.legend(title="Segment")
plt.show()
```

## 8. Çoklu Grafik Paneli: `subplots` ⭐
Bir rapor ya da gösterge paneli (dashboard) hazırlarken birden fazla grafiği tek figürde toplarız. `axes` bir NumPy dizisidir: `axes[satır, sütun]`.

```python
gunler = pd.date_range("2025-01-01", periods=90, freq="D")
gunluk_satis = 200 + np.cumsum(rng.normal(1, 10, 90))

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Satış Gösterge Paneli – 2025 İlk Çeyrek", fontsize=15, fontweight="bold")

axes[0, 0].plot(gunler, gunluk_satis, color="navy")
axes[0, 0].set_title("Günlük Satış")
axes[0, 0].tick_params(axis="x", rotation=30)

axes[0, 1].bar(kategoriler, ciro, color="seagreen")
axes[0, 1].set_title("Kategori Cirosu (bin TL)")
axes[0, 1].tick_params(axis="x", rotation=30)

axes[1, 0].hist(gunluk_satis, bins=15, color="slateblue", edgecolor="white")
axes[1, 0].set_title("Günlük Satış Dağılımı")

axes[1, 1].scatter(calisma_saati, sinav_notu, s=15, alpha=0.6)
axes[1, 1].set_title("Örnek İlişki")

plt.tight_layout()
plt.show()
```

## 10. Korelasyon Isı Haritası (Heatmap)
Veri madenciliğinde değişkenler arası ilişkileri tek bakışta görmek için kullanılır. Çok yüksek korelasyonlu değişkenler modelde **çoklu doğrusal bağlantı** sorununa yol açabilir.

```python
ogr = pd.DataFrame({
    "çalışma_saati": calisma_saati,
    "devam": devam,
    "not": sinav_notu,
    "uyku": rng.normal(7, 1, 120),
    "sosyal_medya": 25 - calisma_saati + rng.normal(0, 3, 120),
})
kor = ogr.corr()

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(kor, cmap="coolwarm", vmin=-1, vmax=1)
fig.colorbar(im, ax=ax, label="Pearson r")
ax.set_xticks(range(len(kor)), kor.columns, rotation=45, ha="right")
ax.set_yticks(range(len(kor)), kor.columns)
for i in range(len(kor)):
    for j in range(len(kor)):
        ax.text(j, i, f"{kor.iloc[i, j]:.2f}", ha="center", va="center",
                color="white" if abs(kor.iloc[i, j]) > 0.5 else "black")
ax.set_title("Öğrenci Değişkenleri Korelasyon Matrisi")
ax.grid(False)
plt.tight_layout()
plt.show()
```

### ✍️ Alıştırma 3 (Kapsamlı): Mini Gösterge Paneli
Aşağıda üretilen `magaza` verisiyle **1×3**'lük bir panel oluşturun:
1. **Sol:** Şubelere göre toplam ciro (sıralı yatay sütun grafiği)
2. **Orta:** Ziyaretçi sayısı ile ciro arasındaki saçılım grafiği, noktalar şubeye göre farklı renkte ve açıklamalı
3. **Sağ:** Şubelere göre günlük ciro dağılımını gösteren kutu grafiği

Panele genel bir başlık (`suptitle`) ekleyin ve `panel.png` olarak kaydedin.

```python
rng_m = np.random.default_rng(11)
subeler = ["Merkez", "Kampüs", "AVM", "Sahil"]
magaza = pd.DataFrame({
    "sube": np.repeat(subeler, 30),
    "ziyaretci": np.concatenate([rng_m.normal(m, 25, 30) for m in [220, 300, 410, 160]]).round(),
})
magaza["ciro"] = (magaza["ziyaretci"] * rng_m.uniform(80, 120, len(magaza))).round()
magaza.head()
```

## 📌 Özet: Hangi Soruya Hangi Grafik?
| Soru | Grafik | Komut |
|---|---|---|
| Zaman içinde nasıl değişti? | Çizgi | `ax.plot` |
| Kategoriler arasında karşılaştırma? | Sütun / yatay sütun | `ax.bar`, `ax.barh` |
| Değişken nasıl dağılmış? | Histogram | `ax.hist` |
| Gruplar arası dağılım, aykırı değer? | Kutu grafiği | `ax.boxplot` |
| İki değişken arasında ilişki var mı? | Saçılım | `ax.scatter` |
| Bütünün parçaları? (az kategori) | Pasta | `ax.pie` |
| Çok değişkenli ilişkiler? | Isı haritası | `ax.imshow` + `corr()` |
