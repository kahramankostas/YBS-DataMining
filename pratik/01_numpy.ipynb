## 1. Neden Liste Değil de NumPy Dizisi?
Python listeleri esnektir ama sayısal hesaplamada yavaştır. NumPy dizileri **tek tip** veri tutar ve işlemleri C dilinde, toplu olarak yapar.

```python
import time

n = 1_000_000
liste = list(range(n))
dizi = np.arange(n)

t0 = time.perf_counter()
sonuc_liste = [x * 2 for x in liste]
t1 = time.perf_counter()
sonuc_dizi = dizi * 2
t2 = time.perf_counter()

print(f"Python listesi : {t1 - t0:.4f} sn")
print(f"NumPy dizisi   : {t2 - t1:.4f} sn")
print(f"NumPy yaklaşık {(t1 - t0) / (t2 - t1):.0f} kat daha hızlı")
```

> ⚠️ **Dikkat:** Listede `*` işlemi tekrar demektir, dizide ise matematiksel çarpmadır!

```python
print([1, 2, 3] * 2)          # liste tekrarı
print(np.array([1, 2, 3]) * 2) # eleman bazında çarpma
```

## 3. İndeksleme ve Dilimleme
İndeksler **0'dan** başlar. Negatif indeks sondan saymayı sağlar. Dilimleme `başlangıç:bitiş:adım` şeklindedir ve bitiş **dahil değildir**.

```python
gunler = np.array(["Pzt", "Sal", "Çar", "Per", "Cum", "Cmt", "Paz"])

print("İlk gün        :", gunler[0])
print("Son gün        :", gunler[-1])
print("Hafta içi      :", gunler[:5])
print("Hafta sonu     :", gunler[5:])
print("Gün atlayarak  :", gunler[::2])
print("Ters sıra      :", gunler[::-1])
```

> ⚠️ **Önemli: Dilim bir kopya değil, görünümdür (view)!** Dilim üzerinde yapılan değişiklik orijinal diziyi de değiştirir. Bağımsız bir kopya için `.copy()` kullanın.

## 4. Boolean Maskeleme (Koşullu Seçim) ⭐
Veri madenciliğinde en çok kullanılan tekniklerden biridir: **bir koşul yazarsınız, NumPy koşulu sağlayan elemanları getirir.**

```python
notlar = np.array([45, 78, 92, 33, 67, 88, 51, 29, 95, 60])

maske = notlar >= 50
print("Maske       :", maske)
print("Geçenler    :", notlar[maske])
print("Geçen sayısı:", maske.sum())   # True = 1, False = 0 olarak toplanır
print("Geçme oranı : %", maske.mean() * 100)
```

```python
# Birden fazla koşul: & (ve), | (veya), ~ (değil). Parantezler ZORUNLUDUR!
print("60–90 arası :", notlar[(notlar >= 60) & (notlar <= 90)])
print("Uç değerler :", notlar[(notlar < 35) | (notlar > 90)])
print("50 altı olmayanlar:", notlar[~(notlar < 50)])
```

```python
# np.where: koşula göre yeni değer atama (Excel'deki EĞER gibi)
durum = np.where(notlar >= 50, "Geçti", "Kaldı")
print(durum)

# Bütünleme sınavına girenlere +10 puan (100'ü geçmesin)
yeni_notlar = np.where(notlar < 50, np.minimum(notlar + 10, 100), notlar)
print(yeni_notlar)
```

## 5. Vektörel İşlemler ve Universal Functions (ufunc)
NumPy'da işlemler **eleman bazındadır**; döngü yazmanıza gerek yoktur.

```python
fiyat = np.array([100, 250, 80, 400])     # TL
adet  = np.array([3, 1, 10, 2])

print("Tutar           :", fiyat * adet)
print("KDV dahil (%20) :", fiyat * 1.20)
print("İndirimli (%15) :", fiyat * (1 - 0.15))
print("Toplam ciro     :", (fiyat * adet).sum())
```

## 6. Toplu İstatistikler ve `axis` Kavramı ⭐
`axis=0` → **sütunlar boyunca** (satırları ezerek, her sütun için bir sonuç)
`axis=1` → **satırlar boyunca** (sütunları ezerek, her satır için bir sonuç)

Kolay hatırlama: `axis` parametresine verdiğiniz eksen **yok olur**.

```python
subeler = np.array(["Gümüşhane", "Trabzon", "Bayburt"])

print("Toplam ciro           :", ciro.sum())
print("Çeyrek toplamları (axis=0):", ciro.sum(axis=0))
print("Şube toplamları  (axis=1):", ciro.sum(axis=1))
print("Şube ortalamaları        :", ciro.mean(axis=1).round(2))
```

## 7. Broadcasting (Yayma) ⭐
Farklı şekillerdeki diziler arasında işlem yapılırken NumPy küçük diziyi otomatik olarak "genişletir". Kural: şekiller sağdan karşılaştırılır; boyutlar **eşit** ya da biri **1** olmalıdır.

```python
# Her çeyreğe farklı enflasyon düzeltmesi uygula: (3,4) * (4,) -> (3,4)
enflasyon_katsayisi = np.array([1.00, 1.05, 1.10, 1.15])
print((ciro * enflasyon_katsayisi).round(2))
```

```python
# Her şubeye farklı hedef: (3,4) - (3,1) -> (3,4)
hedef = np.array([[3.0], [5.0], [2.0]])
print("Hedefe göre fark:\n", ciro - hedef)
```

### Uygulama: Min-Max Normalizasyonu
Veri madenciliğinde değişkenleri aynı ölçeğe getirmek çok önemlidir (özellikle KNN, K-Means gibi mesafe tabanlı algoritmalarda). Min-Max yöntemi değerleri [0, 1] aralığına çeker:

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}}$$

Broadcasting sayesinde bunu tek satırda tüm sütunlar için yapabiliriz.

```python
# Sütunlar: yaş, aylık gelir (TL), yıllık alışveriş sayısı
musteriler = np.array([
    [25, 22000, 12],
    [42, 65000,  4],
    [35, 38000, 25],
    [58, 90000,  2],
    [30, 27000, 18],
])

min_d = musteriler.min(axis=0)
max_d = musteriler.max(axis=0)
normal = (musteriler - min_d) / (max_d - min_d)
print(normal.round(3))
```

## 8. Yeniden Şekillendirme ve Birleştirme

```python
gunluk = np.arange(1, 13)   # 12 aylık satış endeksi
print("Orijinal :", gunluk.shape)

ceyrek = gunluk.reshape(4, 3)       # 4 çeyrek x 3 ay
print("reshape(4,3):\n", ceyrek)

print("reshape(-1, 6) -> -1 'gerisini sen hesapla' demek:", gunluk.reshape(-1, 6).shape)
print("Transpoz (.T) şekli:", ceyrek.T.shape)
print("Düzleştirme (.ravel):", ceyrek.ravel())
```

## 9. Rastgele Sayılar ve Tekrarlanabilirlik

## 10. Eksik Değerler: `np.nan`
Gerçek verilerde eksik değerler olur. NumPy'da eksik değer `np.nan` ile temsil edilir ve normal fonksiyonları "bozar".

```python
sicaklik = np.array([12.5, 14.0, np.nan, 11.2, np.nan, 15.8])

print("Normal ortalama :", sicaklik.mean())        # nan çıkar!
print("nanmean         :", np.nanmean(sicaklik))
print("Eksik sayısı    :", np.isnan(sicaklik).sum())

# Eksikleri ortalama ile doldurma (imputation)
dolu = np.where(np.isnan(sicaklik), np.nanmean(sicaklik), sicaklik)
print("Doldurulmuş     :", dolu.round(2))
```

## 11. Kapanış Uygulaması: Basit Doğrusal Regresyon (Elle)
Reklam harcaması ile satış arasındaki ilişkiyi NumPy ile modelleyelim. En küçük kareler yöntemiyle $y = a \cdot x + b$ doğrusunun katsayıları:
$$a = \frac{\sum (x-\bar{x})(y-\bar{y})}{\sum (x-\bar{x})^2}, \qquad b = \bar{y} - a\bar{x}$$

```python
rng = np.random.default_rng(7)
reklam = rng.uniform(10, 100, size=30)                  # bin TL
satis = 3.2 * reklam + 50 + rng.normal(0, 25, size=30)  # bin TL (gürültülü)

a = ((reklam - reklam.mean()) * (satis - satis.mean())).sum() / ((reklam - reklam.mean()) ** 2).sum()
b = satis.mean() - a * reklam.mean()
print(f"Model: satış = {a:.2f} * reklam + {b:.2f}")

tahmin = a * reklam + b
rmse = np.sqrt(((satis - tahmin) ** 2).mean())
r = np.corrcoef(reklam, satis)[0, 1]
print(f"RMSE: {rmse:.2f}   |   Korelasyon (r): {r:.3f}")
print(f"80 bin TL reklam için tahmini satış: {a * 80 + b:.1f} bin TL")
```
